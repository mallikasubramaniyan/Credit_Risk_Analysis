```python
import numpy as np


days_cols = ['DAYS_FIRST_DUE', 'DAYS_LAST_DUE_1ST_VERSION',
             'DAYS_LAST_DUE', 'DAYS_TERMINATION']
df[days_cols] = df[days_cols].replace(365243, np.nan)


df = df.drop(columns=['RATE_INTEREST_PRIMARY', 'RATE_INTEREST_PRIVILEGED',
                      'DAYS_FIRST_DRAWING'], errors='ignore')


print("Duplicates:", df.duplicated().sum())
df = df.drop_duplicates().reset_index(drop=True)


df['NAME_TYPE_SUITE'] = df['NAME_TYPE_SUITE'].fillna('Unknown')
df['PRODUCT_COMBINATION'] = df['PRODUCT_COMBINATION'].fillna(df['PRODUCT_COMBINATION'].mode()[0])

print(df.shape)
```

```python
ordered_cols = [
    
    'SK_ID_CURR', 'NAME_CONTRACT_TYPE', 'NAME_CLIENT_TYPE', 'NAME_CASH_LOAN_PURPOSE',
    'NAME_PRODUCT_TYPE', 'NAME_PORTFOLIO', 'PRODUCT_COMBINATION', 'NAME_GOODS_CATEGORY',
    
    'NAME_CONTRACT_STATUS', 'CODE_REJECT_REASON', 'DAYS_DECISION',
    
    'AMT_APPLICATION', 'AMT_CREDIT', 'AMT_ANNUITY', 'AMT_GOODS_PRICE',
    'AMT_DOWN_PAYMENT', 'RATE_DOWN_PAYMENT',
    
    'CNT_PAYMENT', 'NAME_YIELD_GROUP', 'NAME_PAYMENT_TYPE', 'NFLAG_INSURED_ON_APPROVAL',
    
    'DAYS_FIRST_DUE', 'DAYS_LAST_DUE_1ST_VERSION', 'DAYS_LAST_DUE', 'DAYS_TERMINATION',
    
    'WEEKDAY_APPR_PROCESS_START', 'HOUR_APPR_PROCESS_START',
    'FLAG_LAST_APPL_PER_CONTRACT', 'NFLAG_LAST_APPL_IN_DAY',
    'CHANNEL_TYPE', 'SELLERPLACE_AREA', 'NAME_SELLER_INDUSTRY', 'NAME_TYPE_SUITE',
]


assert set(ordered_cols) == set(df.columns) and len(ordered_cols) == df.shape[1]
df = df.sort_values('SK_ID_CURR', kind='stable').reset_index(drop=True)

df = df[ordered_cols]
print(df.shape)     
df.head()
```

```python
approved = df[(df['NAME_CONTRACT_STATUS'] == 'Approved') & (df['AMT_CREDIT'] > 0)]
print(approved.shape)   
print(approved[['AMT_APPLICATION', 'AMT_CREDIT', 'AMT_ANNUITY']].describe())

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(approved['AMT_CREDIT'], bins=50, ax=ax[0])
ax[0].set_title("AMT_CREDIT distribution")
sns.boxplot(x=approved['AMT_CREDIT'], ax=ax[1])
ax[1].set_title("AMT_CREDIT outliers")
plt.show()
```

```python
outliers
```

```python
q1 = approved['AMT_CREDIT'].quantile(0.25)
q3 = approved['AMT_CREDIT'].quantile(0.75)
upper = q3 + 1.5 * (q3 - q1)

print("Upper limit:", round(upper))      # 460919
print("Outliers   :", (approved['AMT_CREDIT'] > upper).sum())   # 3422
print("Outlier %  :", round((approved['AMT_CREDIT'] > upper).mean() * 100, 1))   # 10.8
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(approved['AMT_CREDIT'], bins=50, ax=ax[0])
ax[0].set_title("AMT_CREDIT (original)")
sns.histplot(np.log1p(approved['AMT_CREDIT']), bins=50, ax=ax[1])
ax[1].set_title("AMT_CREDIT (log scale)")
plt.show()
```

```python
Correlation, covariance
```

```python
num_cols = ['AMT_APPLICATION', 'AMT_CREDIT', 'AMT_ANNUITY', 'AMT_GOODS_PRICE',
            'CNT_PAYMENT', 'DAYS_DECISION', 'HOUR_APPR_PROCESS_START']

print(approved[num_cols].corr().round(2))
print()
print(approved[['AMT_CREDIT', 'AMT_ANNUITY', 'AMT_GOODS_PRICE']].cov())

plt.figure(figsize=(5, 6))
sns.heatmap(approved[num_cols].corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation (Approved loans)")
plt.show()
```
